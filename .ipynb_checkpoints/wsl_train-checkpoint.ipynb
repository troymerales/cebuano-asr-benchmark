# Bisaya HMM-GMM SAT Training (Kaldi) -- Local WSL2 Edition

A local-WSL rewrite of `kaggle_train.ipynb`, migrating the same pipeline --
HMM-GMM with speaker-adaptive training (SAT), a 2-gram language model, and
the PS27 phoneme set, per `conclusion.md`'s "Conclusions and
Recommendations" (best Bisaya model overall: 2-gram, 3-state, SAT HMM-GMM
using PS27, 5.41% WER, beating every DNN/TDNN variant tested) -- off Kaggle
and onto a local machine running WSL2/Ubuntu. `kaggle_train.ipynb` is left
untouched as the reference for the previous Kaggle-hosted workflow; this
notebook is the active one going forward. See `train.ipynb`'s Section 1 for
the full reasoning behind the pipeline choice -- it isn't repeated here.

This notebook differs from `kaggle_train.ipynb` in what changed to make it
run locally, and nothing else -- the modeling pipeline itself (data split,
lexicon, LM, features, GMM stages, decoding) is unchanged:

1. **No Kaggle environment, paths, or install steps.** `/kaggle/input`,
   `/kaggle/working`, and the Kaggle-session-survival checkpoint
   tar/restore mechanism are all gone -- WSL runs on a persistent local
   disk, so none of that Kaggle-specific machinery is needed.
2. **Kaldi is not cloned.** It's assumed already checked out at `~/kaldi`
   (configurable via the `KALDI_ROOT` environment variable) -- this
   notebook only verifies the checkout and builds `tools/`, KenLM, and
   `src/` against it if that hasn't been done yet.
3. **Dataset and results stay on Windows, referenced through WSL's
   `/mnt/...` bind mount** -- `CORPUS_DIR` and `RESULTS_DIR` are both
   configurable (environment variables, no hardcoded username or drive
   letter). Kaldi's own working directory (`data/`, `mfcc/`, `exp/`, the
   Kaldi checkout itself) stays on the WSL-native filesystem instead,
   since Kaldi's build and training stages create very large numbers of
   small files, and that access pattern is dramatically slower over the
   `/mnt/...` Windows bind mount than on native ext4. See
   `LOCAL_WSL_SETUP.md` for the full rationale and setup instructions.

**Why data loading looks different from `kagglebasis.ipynb`:** unchanged
from `kaggle_train.ipynb` -- see that notebook's intro cell for the
pyarrow/nested-column details. This notebook keeps the same per-shard
`pd.read_parquet()` + in-memory `pd.concat()` approach (Section 7).

**Environment requirement:** Kaldi does not build on native Windows. This
notebook must run inside WSL2/Ubuntu (Kaggle is no longer part of this
project's workflow).

## 2. Configuration

`CORPUS_DIR` points at the dataset on the Windows filesystem (accessed
through WSL's `/mnt/...` bind mount) -- configurable via the
`BISAYA_CORPUS_DIR` environment variable, defaulting to `data/bisaya_audio`
next to this notebook (the same relative path `kaggle_train.ipynb` already
fell back to outside Kaggle). The corpus is **not** pre-split into
train/test shards -- Section 7/8 load every shard and derive a
speaker-independent split in memory, so there is no separate
"split-generation" step and nothing gets written back out as a recombined
Parquet file.

`WORK_DIR` holds everything this notebook creates (`data/`, `mfcc/`,
`exp/`, the recipe scaffolding) and defaults to a directory under `$HOME`
-- i.e. the WSL-native filesystem, not a Windows-mounted path. Kaldi's
build and training stages create very large numbers of small files, and
WSL2's `/mnt/...` bind mount is dramatically slower than native ext4 for
that access pattern, so keeping `WORK_DIR` on the WSL side is a deliberate
performance choice, not an oversight -- see `LOCAL_WSL_SETUP.md`'s
"Storage considerations" section. Final results are copied back out to
`RESULTS_DIR` on Windows in Section 15. Unlike `kaggle_train.ipynb`, there
is no tar/untar checkpoint mechanism here -- WSL's disk is persistent
across sessions on its own, so `stage()` (Section 2's other cell) provides
all the resumability that's needed.

```python
import os
import subprocess
from pathlib import Path

# WORK_DIR holds everything this notebook creates (Kaldi recipe
# scaffolding, data/, mfcc/, exp/). Kaldi's build and training stages
# create very large numbers of small files (object files while compiling;
# then per-utterance .ark/.scp shards, alignment logs, lattices while
# training/decoding). On WSL2, paths under /mnt/... are bridged to the
# Windows NTFS volume via 9P/DrvFs, which is dramatically slower than
# WSL2's native ext4 for exactly this many-small-file pattern -- so unlike
# the dataset/results below, WORK_DIR intentionally defaults to somewhere
# under $HOME (native WSL filesystem), not a /mnt/... path. See
# LOCAL_WSL_SETUP.md's "Storage considerations" section. Only the final
# model/decode results (Section 15) get copied back out to Windows.
WORK_DIR = Path(os.environ.get("BISAYA_WORK_DIR", str(Path.home() / "bisaya_asr"))).resolve()
WORK_DIR.mkdir(parents=True, exist_ok=True)

# Kaldi is assumed already cloned locally -- this notebook never runs
# `git clone` for it. Defaults to ~/kaldi; override KALDI_ROOT if yours
# lives elsewhere.
KALDI_ROOT = Path(os.environ.get("KALDI_ROOT", str(Path.home() / "kaldi"))).resolve()
assert KALDI_ROOT.exists(), (
    f"KALDI_ROOT={KALDI_ROOT} does not exist. This notebook expects an "
    f"already-cloned Kaldi checkout (per the migration brief, at ~/kaldi) "
    f"-- set the KALDI_ROOT environment variable if yours lives elsewhere. "
    f"This notebook does not clone Kaldi itself."
)

DATA_ROOT = WORK_DIR / "data"
MFCC_ROOT = WORK_DIR / "mfcc"
EXP_ROOT = WORK_DIR / "exp"

# The dataset stays on Windows and is read directly through WSL's /mnt/...
# bind mount -- same relative path ("data/bisaya_audio", next to this
# notebook) kaggle_train.ipynb already fell back to whenever /kaggle
# wasn't present. Reading a handful of large Parquet shards sequentially
# (Section 7) isn't the many-small-file pattern that makes /mnt/... slow,
# so this is fine to leave on Windows. Override BISAYA_CORPUS_DIR if your
# corpus lives somewhere else, e.g. a dedicated data drive:
#   export BISAYA_CORPUS_DIR=/mnt/d/ASR/dataset/bisaya_audio
CORPUS_DIR = Path(os.environ.get("BISAYA_CORPUS_DIR", "data/bisaya_audio")).resolve()

# Final artifacts (trained model, decode graph, WER results -- Section 15)
# are copied back to a Windows-hosted directory here so they're available
# outside WORK_DIR/this WSL distro. Defaults to an output/ folder next to
# this notebook; override BISAYA_RESULTS_DIR to put it elsewhere.
RESULTS_DIR = Path(os.environ.get("BISAYA_RESULTS_DIR", "output")).resolve()

print(f"WORK_DIR    = {WORK_DIR}")
print(f"KALDI_ROOT  = {KALDI_ROOT}")
print(f"CORPUS_DIR  = {CORPUS_DIR}")
print(f"RESULTS_DIR = {RESULTS_DIR}")
if not CORPUS_DIR.exists():
    print(f"WARNING: CORPUS_DIR does not exist -- set BISAYA_CORPUS_DIR to "
          f"your actual dataset path before running Section 7.")
```

```python
def sh(cmd, cwd=None, check=True, env=None):
    # Streams output live -- used for every Kaldi binary/script invocation
    # below instead of `!` magics, so it behaves the same interactively or
    # when the notebook is replayed top to bottom.
    print(f"$ {cmd}")
    proc = subprocess.run(cmd, shell=True, cwd=cwd, check=check, executable="/bin/bash", env=env)
    return proc.returncode


def stage(name, done, fn):
    # Generic resumability helper used by every expensive step in this
    # notebook. `done` is a zero-arg callable returning True if this
    # stage's output already exists; `fn` does the actual work. Keeps
    # re-running this notebook top-to-bottom cheap after an interrupted
    # kernel or machine restart -- WORK_DIR lives on local disk and
    # persists on its own, so (unlike the Kaggle notebook this was
    # migrated from) there's no separate tar/untar checkpoint step needed
    # -- only whatever stage didn't finish actually re-runs.
    if done():
        print(f"[skip] {name}: output already exists")
        return
    print(f"[run]  {name}")
    fn()
```

## 3. Install Dependencies

Ubuntu package installation via `apt-get` -- the same set the pipeline
needed under Kaggle's Debian-based image. This requires `sudo` on a normal
WSL Ubuntu install (Kaggle's container ran as root, so its version of this
cell didn't need it). `sudo apt-get` from inside a notebook's subprocess
will hang waiting for a password prompt if your sudo timestamp isn't
already cached -- if this cell seems stuck, run the same `apt-get install`
line once from a WSL terminal first (which prompts for your password
normally), then re-run this cell; it will find everything already
installed and skip straight through.

Skipped entirely on a re-run once `.deps_installed` exists under
`WORK_DIR` -- `apt-get`/`pip` are themselves idempotent, but skipping the
call outright avoids the overhead of `apt-get update` on every re-run.

```python
import shutil

DEPS_MARKER = WORK_DIR / ".deps_installed"

def _install_apt_deps():
    #sh("sudo apt-get update -qq && sudo apt-get install -y -qq "
    #   "build-essential automake autoconf libtool subversion git zlib1g-dev "
    #   "gfortran sox perl cmake libboost-all-dev libeigen3-dev")
    DEPS_MARKER.touch()

stage("apt dependencies", DEPS_MARKER.exists, _install_apt_deps)

# The checks below are separate from the marker above (which may already
# exist from an earlier run that predates one of these being added to the
# list) -- each is checked and installed independently so a resumed
# WORK_DIR still gets it even if the main install was skipped entirely.
#
# perl: Kaldi's steps/utils scripts are Perl; missing it fails with a bare
# "exit status 127" wherever the first .pl script runs.
stage(
    "install perl (if missing)",
    lambda: shutil.which("perl") is not None,
    lambda: sh("sudo apt-get install -y -qq perl"),
)

# cmake: checked by absolute apt install path (/usr/bin/cmake), NOT
# shutil.which("cmake") -- a pip-installed cmake on PATH ahead of /usr/bin
# would make shutil.which find that one instead and consider this stage
# already done. Pinning to apt's cmake via absolute path (Section 4 does
# the same) keeps the toolchain deterministic.
stage(
    "install cmake (if missing)",
    lambda: Path("/usr/bin/cmake").exists(),
    lambda: sh("sudo apt-get install -y -qq cmake"),
)

# boost: KenLM's own CMake build (Section 4 -- our direct build, not
# Kaldi's restricted install_kenlm_query_only.sh) needs Boost to produce
# lmplz. libboost-all-dev is heavier than a targeted subset, but avoids
# yet another single-missing-component 127 after everything else built.
stage(
    "install boost + eigen (if missing)",
    lambda: Path("/usr/include/boost/version.hpp").exists(),
    lambda: sh("sudo apt-get install -y -qq libboost-all-dev libeigen3-dev"),
)


def _pip_deps_present():
    try:
        import pandas, pyarrow, tqdm  # noqa: F401
        return True
    except ImportError:
        return False

stage(
    "pip dependencies (pandas, pyarrow, tqdm)",
    _pip_deps_present,
    lambda: sh("pip install -q pandas pyarrow tqdm"),
)
```

## 4. Kaldi: Verify Existing Checkout, Build Tools + KenLM

`KALDI_ROOT` (`~/kaldi` by default) is assumed already cloned -- this
notebook never runs `git clone` for Kaldi itself, only checks that the
checkout looks complete. The `tools/` build (OpenFST etc.) and KenLM
install are each skipped once their own output binary/library is found,
rather than re-invoking `make`/the installer script every time.

```python
def _kaldi_checkout_complete():
    canaries = [
        KALDI_ROOT / "tools" / "extras" / "install_kenlm_query_only.sh",
        KALDI_ROOT / "egs" / "wsj" / "s5" / "steps",
        KALDI_ROOT / "egs" / "wsj" / "s5" / "utils",
        KALDI_ROOT / "src" / "Makefile",
    ]
    return all(c.exists() for c in canaries)

assert _kaldi_checkout_complete(), (
    f"{KALDI_ROOT} exists but doesn't look like a complete Kaldi checkout "
    f"(missing one of tools/extras/install_kenlm_query_only.sh, "
    f"egs/wsj/s5/steps, egs/wsj/s5/utils, src/Makefile). This notebook "
    f"never clones Kaldi -- fix or re-clone the checkout at {KALDI_ROOT} "
    f"manually, or point KALDI_ROOT at a complete one."
)
print(f"Using existing Kaldi checkout at {KALDI_ROOT}")


# Capped well below os.cpu_count(): Kaldi/KenLM/sph2pipe compile heavy C++
# translation units (cc1plus RSS over 1.5 GiB each is normal for Kaldi's
# src/), and `make -j$(nproc)` on an 8-core/7.6 GiB-RAM WSL2 VM has been
# observed to OOM-kill the compiler mid-build. Drop to 1 below if 2 still
# causes memory pressure -- GNU Make resumes an interrupted build from
# wherever it left off (already-built objects aren't recompiled), so
# lowering this and re-running is always safe, no clean/reclone needed.
MAKE_JOBS = 2

def _build_tools():
    sh(f"make -j{MAKE_JOBS} CCFLAGS=-std=gnu17", cwd=KALDI_ROOT / "tools")

stage(
    "build Kaldi tools (OpenFST etc.)",
    lambda: (KALDI_ROOT / "tools" / "openfst" / "lib").exists(),
    _build_tools,
)

def _openblas_root():
    return KALDI_ROOT / "tools" / "OpenBLAS" / "install"


def _install_openblas():
    # Kaldi's script downloads+extracts OpenBLAS 0.3.13, then runs its
    # own `make ... all install` -- which fails to compile under GCC 14+'s
    # newer default-error diagnostics: -Wincompatible-pointer-types (hit in
    # the AVX-512 kernel code) and -Wimplicit-function-declaration (hit in
    # OpenBLAS's own linktest sanity check) both flipped from warning to
    # hard error by default starting around GCC 14. A `make CC=...` override
    # can't be threaded through the stock script, so its download/extract
    # runs first (tolerating its now-expected failure at the make step via
    # check=False), then the actual build+install re-runs here with the fix
    # applied -- against the same already-extracted OpenBLAS/ dir, so make's
    # normal incremental behavior means only the previously-failing files
    # actually get recompiled.
    sh("extras/install_openblas.sh", cwd=KALDI_ROOT / "tools", check=False)
    openblas_src = KALDI_ROOT / "tools" / "OpenBLAS"
    cc_override = (
        "gcc -Wno-error=implicit-function-declaration "
        "-Wno-error=incompatible-pointer-types -Wno-error=int-conversion"
    )
    sh(f'make CC="{cc_override}" PREFIX={openblas_src}/install '
       f'USE_LOCKING=1 USE_THREAD=0 -C {openblas_src} all install')

stage(
    "build OpenBLAS (Kaldi's math library)",
    lambda: (_openblas_root() / "lib" / "libopenblas.so").exists(),
    _install_openblas,
)

def _find_lmplz():
    # Prefer an already-installed system lmplz over building a redundant
    # copy inside the Kaldi checkout.
    system_lmplz = shutil.which("lmplz")
    if system_lmplz:
        return Path(system_lmplz)
    local_lmplz = KALDI_ROOT / "tools" / "kenlm" / "build" / "bin" / "lmplz"
    return local_lmplz if local_lmplz.exists() else None


def _install_kenlm():
    # Kaldi's own tools/extras/install_kenlm_query_only.sh deliberately
    # does NOT build lmplz -- straight from that script's header: a
    # full build (with arpa counting/smoothing/interpolation support)
    # depends on Eigen and Boost, "too heavy to integrate" there. Section
    # 11 needs lmplz to build the 2-gram ARPA, so KenLM is built directly
    # here instead, inside the existing Kaldi checkout's tools/ dir --
    # not a separate, duplicate installation.
    #
    # -DBoost_NO_BOOST_CMAKE=ON works around a Debian/Ubuntu packaging gap:
    # libboost-all-dev ships Boost's umbrella CMake config file but not the
    # per-component ones (boost_program_optionsConfig.cmake etc.) that
    # CMake's config-mode search then needs, so it fails deeper in instead
    # of ever falling back to a plain library/header search. This flag
    # skips Boost's CMake package entirely and forces that plain search,
    # which succeeds since libboost-all-dev does provide the actual .so
    # files and headers, just not those specific per-component CMake
    # config files.
    kenlm_dir = KALDI_ROOT / "tools" / "kenlm"
    sh(f"rm -rf {kenlm_dir}")
    sh(f"git clone https://github.com/kpu/kenlm.git {kenlm_dir}")

    # KenLM's CMakeLists.txt requires a "system" Boost component --
    # Boost.System has been header-only for years, and this Ubuntu's Boost
    # (1.90) has finally dropped the compiled compatibility stub entirely,
    # so no libboost_system.* exists anywhere to find, and CMake's
    # find_package(Boost REQUIRED COMPONENTS ... system ...) fails outright
    # ("Could NOT find Boost (missing: system)") even though every other
    # requested component (program_options, thread, unit_test_framework) is
    # present. Nothing in KenLM actually needs a *compiled* Boost.System --
    # dropping it from the requested component list is enough; confirmed
    # this still configures and builds cleanly.
    sh(f"sed -i '/^  system$/d' {kenlm_dir}/CMakeLists.txt")

    sh("mkdir -p build && cd build && "
       "/usr/bin/cmake -DBoost_NO_BOOST_CMAKE=ON .. && "
       f"make -j{MAKE_JOBS}", cwd=kenlm_dir)

stage(
    "install KenLM (full build, for lmplz)",
    lambda: _find_lmplz() is not None,
    _install_kenlm,
)
```

## 5. Kaldi Recipe: Build Kaldi Binaries (`src/`)

Skipped entirely if `compute-mfcc-feats` already exists -- this is the
slowest step in the whole notebook (30-90 minutes cold), so avoiding a
redundant `make` invocation (even an incremental no-op one) matters most
here.

```python
src_dir = KALDI_ROOT / "src"
mfcc_bin = src_dir / "featbin" / "compute-mfcc-feats"  # featbin/, not bin/ -- see LOCAL_WSL_SETUP.md

# use-cuda=no: targets CPU-only environments (no CUDA-capable GPU assumed).
# If your machine has a CUDA-capable GPU you want Kaldi to use, drop
# --use-cuda=no.
def _build_kaldi_src():
    if not (src_dir / "kaldi.mk").exists():
        sh(f"./configure --shared --use-cuda=no --mathlib=OPENBLAS "
           f"--openblas-root={_openblas_root()}", cwd=src_dir)
    sh(f"make -j{MAKE_JOBS} depend", cwd=src_dir)
    sh(f"make -j{MAKE_JOBS}", cwd=src_dir)

stage("build Kaldi (src/)", mfcc_bin.exists, _build_kaldi_src)

print("Kaldi build complete." if mfcc_bin.exists()
      else "WARNING: expected binary not found -- check the build log above for errors.")
```

## 6. Kaldi Recipe Scaffolding (`path.sh`, `cmd.sh`, `steps/`, `utils/`)

Writing `path.sh`/`cmd.sh` and symlinking `steps/`/`utils/` is already cheap
and idempotent (the symlink check below skips re-linking), so this isn't
wrapped in `stage()`.

```python
path_sh_lines = [
    f"export KALDI_ROOT={KALDI_ROOT}",
    "export PATH=$PWD/utils/:$KALDI_ROOT/tools/openfst/bin:$PWD:$PATH",
    ". $KALDI_ROOT/tools/config/common_path.sh",
    "export LC_ALL=C",
]
(WORK_DIR / "path.sh").write_text("\n".join(path_sh_lines) + "\n")

cmd_sh_lines = [
    "export train_cmd=run.pl",
    "export decode_cmd=run.pl",
    "export mkgraph_cmd=run.pl",
]
(WORK_DIR / "cmd.sh").write_text("\n".join(cmd_sh_lines) + "\n")

wsj_s5 = KALDI_ROOT / "egs" / "wsj" / "s5"
for name in ("steps", "utils"):
    link = WORK_DIR / name
    if not link.exists():
        link.symlink_to(wsj_s5 / name)

# steps/make_mfcc.sh (Section 12) requires conf/mfcc.conf by default --
# every egs/*/s5 recipe ships one checked in, but this WORK_DIR isn't a
# real recipe directory (only steps/ and utils/ are symlinked in above),
# so it has to be created explicitly. Same content as Kaldi's own wsj s5
# recipe -- the only non-default MFCC option this pipeline uses.
(WORK_DIR / "conf").mkdir(exist_ok=True)
mfcc_conf = WORK_DIR / "conf" / "mfcc.conf"
if not mfcc_conf.exists():
    mfcc_conf.write_text("--use-energy=false   # only non-default option.\n")

# Sanity check: a broken symlink (e.g. Kaldi's src/ build didn't finish, or
# KALDI_ROOT points somewhere unexpected) makes every steps/utils/*.pl call
# fail later with a bare, confusing "exit status 127" ("command not
# found") -- catching it here instead gives an immediate, specific error.
probe = WORK_DIR / "utils" / "utt2spk_to_spk2utt.pl"
assert probe.exists(), (
    f"{probe} not found -- steps/utils symlinks are broken (check that "
    f"{wsj_s5} exists and KALDI_ROOT={KALDI_ROOT} matches the Kaldi checkout "
    f"these were linked from)."
)

print("path.sh, cmd.sh, steps/, utils/ ready under", WORK_DIR)
```

## 7. Data Loading (Windows-Hosted Corpus, No Intermediate Parquet Rewrite)

Reads every shard in `CORPUS_DIR` straight into pandas with
`pd.read_parquet()` and concatenates in memory -- no `pa.concat_tables()` +
`pq.write_table()` round trip (see this notebook's intro cell for why that
combination fails for a nested `audio` column on at least one pyarrow
build this pipeline has run against). Each shard is read exactly once,
exactly the way it already exists on disk -- `CORPUS_DIR` can point
straight at the Windows-hosted dataset through `/mnt/...` with no copy
step, since this is a handful of large sequential reads, not the
many-small-file pattern that's slow over that mount.

```python
del corpus_df, train_df, test_df
import gc
gc.collect()
```

```python
import pandas as pd

def load_all_shards(corpus_dir):
    files = sorted(Path(corpus_dir).glob("*.parquet"))
    if not files:
        return pd.DataFrame()
    dfs = []
    for file in files:
        print(f"Loading: {file.name}")
        temp = pd.read_parquet(file)
        temp["source_file"] = file.name
        dfs.append(temp)
    return pd.concat(dfs, ignore_index=True)


corpus_df = load_all_shards(CORPUS_DIR)
print(f"\nLoaded {len(corpus_df):,} utterances from {CORPUS_DIR}")
print(f"Speakers: {corpus_df['speaker_id'].nunique() if len(corpus_df) else 0:,}")
```

## 8. Speaker-Independent Train/Test Split + Kaldi Data Dirs

Split **by speaker**, not by utterance, ~80/20, with a fixed seed (printed
below) so it's reproducible on rerun -- matches `train.ipynb` Section 6.
`build_kaldi_data_dir()` (which writes every utterance's audio out as an
individual `.wav` file plus `wav.scp`/`text`/`utt2spk`) is skipped per split
if that split's `wav.scp` already exists, since re-writing thousands of wav
files on every re-run is exactly the kind of redundant work this notebook
is meant to avoid. These per-utterance `.wav` files land under `WORK_DIR`
(WSL-native), not `CORPUS_DIR` -- writing thousands of small files is
exactly the access pattern that's slow over `/mnt/...`.

```python
import re

SPLIT_SEED = 42  # fixed and printed so this split is reproducible on rerun

def speaker_independent_split(df, test_fraction=0.2, seed=SPLIT_SEED):
    speakers = sorted(df["speaker_id"].unique())
    shuffled = pd.Series(speakers).sample(frac=1.0, random_state=seed)
    n_test = max(1, round(len(speakers) * test_fraction))
    test_speakers = set(shuffled.iloc[:n_test])
    train_speakers = set(shuffled.iloc[n_test:])

    assert train_speakers.isdisjoint(test_speakers)

    return df[df["speaker_id"].isin(train_speakers)].copy(), df[df["speaker_id"].isin(test_speakers)].copy()


train_df, test_df = speaker_independent_split(corpus_df)

print(f"seed = {SPLIT_SEED}")
print(f"train: {train_df['speaker_id'].nunique()} speakers, {len(train_df)} utterances")
print(f"test:  {test_df['speaker_id'].nunique()} speakers, {len(test_df)} utterances")
print(f"speaker overlap: {set(train_df['speaker_id']) & set(test_df['speaker_id'])}")


def kaldi_normalize_text(text):
    # Lowercase + strip punctuation. Deliberately does NOT collapse u/o the
    # way main.ipynb's normalize_bisaya does for WER scoring -- that merge
    # is a scoring-time leniency, not a valid training transcript for a
    # model that's supposed to learn the u/o distinction from audio.
    text = text.lower()
    text = re.sub(r"[^\w\s']", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text


def build_kaldi_data_dir(df, out_dir, wav_out_dir):
    out_dir = Path(out_dir)
    wav_out_dir = Path(wav_out_dir)

    if (out_dir / "wav.scp").exists():
        print(f"[skip] {out_dir}: wav.scp already exists")
        return

    out_dir.mkdir(parents=True, exist_ok=True)
    wav_out_dir.mkdir(parents=True, exist_ok=True)

    wav_lines, text_lines, utt2spk_lines = [], [], []

    for i, row in df.iterrows():
        # utt-ids are prefixed with speaker-id -- Kaldi's sort-order
        # conventions require this for utt2spk/spk2utt to line up.
        spk = str(row["speaker_id"])
        utt_id = f"{spk}-{i:06d}"

        wav_path = wav_out_dir / f"{utt_id}.wav"
        wav_path.write_bytes(row["audio"]["bytes"])

        # Piped through sox so every utterance ends up at one consistent
        # sample rate/channel count regardless of the source file's own rate.
        wav_lines.append(f"{utt_id} sox {wav_path} -r 16000 -c 1 -t wav - |")

        transcript = kaldi_normalize_text(str(row["transcript"]))
        text_lines.append(f"{utt_id} {transcript}")
        utt2spk_lines.append(f"{utt_id} {spk}")

    (out_dir / "wav.scp").write_text("\n".join(sorted(wav_lines)) + "\n")
    (out_dir / "text").write_text("\n".join(sorted(text_lines)) + "\n")
    (out_dir / "utt2spk").write_text("\n".join(sorted(utt2spk_lines)) + "\n")

    sh(f"utils/utt2spk_to_spk2utt.pl {out_dir}/utt2spk > {out_dir}/spk2utt", cwd=WORK_DIR)
    sh(f"utils/fix_data_dir.sh {out_dir}", cwd=WORK_DIR)
    # --non-print: this Ubuntu only has the C.utf8 locale installed, but
    # this script specifically looks for "C.UTF-8" or "en_US.UTF-8" (neither
    # present) for its non-printable-character check, silently falls back to
    # the plain C locale when it finds neither, and then misclassifies
    # ordinary non-ASCII UTF-8 bytes (accented Bisaya characters) as
    # "non-printable" -- a false positive, confirmed by re-running the same
    # check with the locale that's actually installed (0 flagged lines).
    sh(f"utils/validate_data_dir.sh --no-feats --non-print {out_dir}", cwd=WORK_DIR)

    print(f"{out_dir}: {len(wav_lines)} utterances, {df['speaker_id'].nunique()} speakers")


build_kaldi_data_dir(train_df, DATA_ROOT / "train", WORK_DIR / "wav" / "train")
build_kaldi_data_dir(test_df, DATA_ROOT / "test", WORK_DIR / "wav" / "test")

# corpus_df/train_df/test_df hold the entire corpus's raw audio bytes in
# memory (this repo's Parquet shards total ~1.9 GB compressed, several
# times that once decoded) -- dead weight from here on, since every
# utterance has already been written out to its own .wav file above. On a
# resource-constrained local machine (unlike Kaggle's ~30 GB-RAM notebooks)
# leaving these alive for the rest of the session can consume most of
# available RAM, causing later stages (MFCC extraction, GMM training) to
# fail with `OSError: [Errno 12] Cannot allocate memory` when they try to
# spawn subprocesses with little headroom left.
```

```python
del corpus_df, train_df, test_df
import gc
gc.collect()
```

## 10. Lexicon and Phoneme Data (`data/local/dict`, PS27)

Builds a lexicon constrained to **PS27**'s 27-monophone inventory, exactly
as in `train.ipynb` Section 8 -- see that notebook for the full rationale.
Lexicon build and `prepare_lang.sh` are each skipped if their own output
already exists.

```python
PS27_PHONES = {
    "p", "b", "t", "d", "k", "g",
    "f", "v", "s", "z", "sh", "th", "h",
    "j", "ch",
    "m", "n", "ng",
    "l", "r",
    "w", "y",
    "a", "e", "i", "o", "u",
}


def word_to_phones(word):
    w = re.sub(r"[^a-z']", "", word.lower()).replace("'", "")
    phones = []
    i = 0
    while i < len(w):
        if w[i:i + 2] == "ng":
            phones.append("ng")
            i += 2
        elif w[i:i + 2] == "sh":
            phones.append("sh")
            i += 2
        elif w[i:i + 2] == "th":
            phones.append("th")
            i += 2
        elif w[i:i + 2] in ("ts", "ty"):
            phones.append("ch")
            i += 2
        elif w[i] == "c":
            phones.append("s" if w[i + 1:i + 2] in ("e", "i") else "k")
            i += 1
        elif w[i] == "q":
            phones.append("k")
            i += 1
        elif w[i] == "x":
            phones.extend(["k", "s"])
            i += 1
        else:
            phones.append(w[i])
            i += 1

    assert all(p in PS27_PHONES for p in phones), f"{word} -> {phones} outside PS27"
    return phones


def build_lexicon(text_paths, dict_dir):
    dict_dir = Path(dict_dir)

    if (dict_dir / "lexicon.txt").exists():
        print(f"[skip] {dict_dir}: lexicon.txt already exists")
        return dict_dir

    dict_dir.mkdir(parents=True, exist_ok=True)

    vocab = set()
    for p in text_paths:
        for line in Path(p).read_text().splitlines():
            words = line.split(" ")[1:]
            vocab.update(w for w in words if w)

    lexicon_lines = ["<unk> spn"]
    all_phones = set()
    for word in sorted(vocab):
        phones = word_to_phones(word)
        if not phones:
            continue
        all_phones.update(phones)
        lexicon_lines.append(f"{word} {' '.join(phones)}")

    (dict_dir / "lexicon.txt").write_text("\n".join(lexicon_lines) + "\n")
    (dict_dir / "silence_phones.txt").write_text("sil\nspn\n")
    (dict_dir / "optional_silence.txt").write_text("sil\n")
    (dict_dir / "nonsilence_phones.txt").write_text("\n".join(sorted(all_phones)) + "\n")
    (dict_dir / "extra_questions.txt").write_text("")

    print(f"Vocabulary: {len(vocab)} words, {len(all_phones)} distinct phones (of PS27's 27) -> {dict_dir}")
    return dict_dir


LOCAL_DICT_DIR = build_lexicon(
    [DATA_ROOT / "train" / "text", DATA_ROOT / "test" / "text"],
    DATA_ROOT / "local" / "dict",
)

LANG_DIR = DATA_ROOT / "lang"

def _prepare_lang():
    sh(f"utils/prepare_lang.sh --position-dependent-phones false "
       f"{LOCAL_DICT_DIR} '<unk>' {DATA_ROOT}/local/lang {LANG_DIR}", cwd=WORK_DIR)

stage("prepare_lang", lambda: (LANG_DIR / "L.fst").exists(), _prepare_lang)
```

## 11. Language Model (KenLM -> ARPA -> Kaldi `G.fst`)

Builds a **2-gram** word-level LM from the training transcripts -- matches
`train.ipynb` Section 9 (`conclusion.md` reports 2-gram as best for Bisaya).
Skipped if `lang_2g/G.fst` already exists. Uses whichever `lmplz` Section
4 found or built (`_find_lmplz()`), so this doesn't care whether that came
from a pre-existing system install or the KenLM build inside the Kaldi
checkout.

```python
LM_DIR = DATA_ROOT / "local" / "lm"
LANG_2G_DIR = DATA_ROOT / "lang_2g"

def _build_lm():
    lmplz = _find_lmplz()
    assert lmplz is not None, "lmplz not found -- Section 4's KenLM install stage should have built it."
    LM_DIR.mkdir(parents=True, exist_ok=True)

    train_text_lines = (DATA_ROOT / "train" / "text").read_text().splitlines()
    corpus_txt = LM_DIR / "corpus.txt"
    corpus_txt.write_text("\n".join(" ".join(line.split(" ")[1:]) for line in train_text_lines) + "\n")

    arpa_path = LM_DIR / "2gram.arpa"
    sh(f"{lmplz} -o 2 --discount_fallback < {corpus_txt} > {arpa_path}")

    # utils/format_lm.sh unconditionally runs `gunzip -c` on the ARPA file
    # regardless of its extension -- it assumes an already-gzipped LM (the
    # Kaldi convention, e.g. from IRSTLM/SRILM), but lmplz writes plain text.
    # Without this, format_lm.sh fails with "not in gzip format" and feeds
    # arpa2fst an empty stream ("\data\ section missing or empty").
    arpa_gz_path = Path(f"{arpa_path}.gz")
    sh(f"gzip -kf {arpa_path}")

    sh(f"utils/format_lm.sh {LANG_DIR} {arpa_gz_path} {LOCAL_DICT_DIR}/lexicon.txt "
       f"{LANG_2G_DIR}", cwd=WORK_DIR)

stage("build 2-gram LM", lambda: (LANG_2G_DIR / "G.fst").exists(), _build_lm)

print("lang_2g graph directory:", LANG_2G_DIR)
```

25 ms window, 10 ms frameshift, 13 static coefficients -- Kaldi's MFCC
defaults already match this. CMVN is per-speaker. Matches `train.ipynb`
Section 10. Skipped per split if that split's `feats.scp` already exists
and is non-empty.

```python
N_JOBS = min(8, os.cpu_count() or 4)

def _make_extract_fn(name, data_dir):
    def _fn():
        sh(f"steps/make_mfcc.sh --cmd run.pl --nj {N_JOBS} {data_dir} {EXP_ROOT}/make_mfcc/{name} {MFCC_ROOT}", cwd=WORK_DIR)
        sh(f"steps/compute_cmvn_stats.sh {data_dir} {EXP_ROOT}/make_mfcc/{name} {MFCC_ROOT}", cwd=WORK_DIR)
        sh(f"utils/fix_data_dir.sh {data_dir}", cwd=WORK_DIR)
    return _fn


for name, data_dir in [("train", TRAIN_DATA_DIR), ("test", DATA_ROOT / "test")]:
    feats_scp = Path(data_dir) / "feats.scp"
    stage(
        f"MFCC+CMVN ({name})",
        lambda p=feats_scp: p.exists() and p.stat().st_size > 0,
        _make_extract_fn(name, data_dir),
    )
```

## 13. HMM-GMM Training (Monophone -> Triphone -> LDA+MLLT -> SAT)

The standard Kaldi progression: monophone, then triphone (delta features),
then LDA+MLLT, then speaker-adaptive training (SAT) -- SAT is this
notebook's **final model**, matching `train.ipynb` Section 11. Each stage is
skipped if its own `final.mdl` already exists.

```python
# Alignment/retry beams widened well past Kaldi's WSJ-recipe defaults
# (initial_beam=6, regular_beam=10, retry_beam=40 in train_mono.sh; beam=10,
# retry_beam=40 in train_deltas.sh/train_lda_mllt.sh/train_sat.sh). Those
# defaults assume short read-speech sentences (a few seconds each); this
# corpus's utterances average ~105s and run up to ~330s, which caused ~89%
# of monophone alignments to fail outright at the defaults (confirmed by
# direct testing). Widening to this still doesn't reach 100% -- a handful
# of utterances (long code-switched sentences in particular, where the
# PS27 G2P lexicon's Bisaya letter-to-phone rules poorly approximate the
# actual English pronunciation used) fail regardless of beam width -- but
# it recovers the large majority that were failing purely from beam
# pruning on long utterances, not a genuine mismatch. If per-iteration
# alignment failure counts (steps/train_mono.sh log) don't keep dropping
# after the first couple of iterations, that's a sign this corpus's
# longest utterances need to be split into shorter segments rather than
# pushing the beam even wider.
ALIGN_BEAM = 40
ALIGN_RETRY_BEAM = 160


def _train_mono():
    sh(f"steps/train_mono.sh --cmd run.pl --nj {N_JOBS} "
       f"--initial-beam {ALIGN_BEAM} --regular-beam {ALIGN_BEAM} --retry-beam {ALIGN_RETRY_BEAM} "
       f"{TRAIN_DATA_DIR} {LANG_DIR} {EXP_ROOT}/mono", cwd=WORK_DIR)
    sh(f"steps/align_si.sh --cmd run.pl --nj {N_JOBS} "
       f"--beam {ALIGN_BEAM} --retry-beam {ALIGN_RETRY_BEAM} "
       f"{TRAIN_DATA_DIR} {LANG_DIR} {EXP_ROOT}/mono {EXP_ROOT}/mono_ali", cwd=WORK_DIR)

stage("monophone (mono)", lambda: (EXP_ROOT / "mono" / "final.mdl").exists(), _train_mono)


def _train_tri1():
    sh(f"steps/train_deltas.sh --cmd run.pl "
       f"--beam {ALIGN_BEAM} --retry-beam {ALIGN_RETRY_BEAM} 2000 10000 "
       f"{TRAIN_DATA_DIR} {LANG_DIR} {EXP_ROOT}/mono_ali {EXP_ROOT}/tri1", cwd=WORK_DIR)
    sh(f"steps/align_si.sh --cmd run.pl --nj {N_JOBS} "
       f"--beam {ALIGN_BEAM} --retry-beam {ALIGN_RETRY_BEAM} "
       f"{TRAIN_DATA_DIR} {LANG_DIR} {EXP_ROOT}/tri1 {EXP_ROOT}/tri1_ali", cwd=WORK_DIR)

stage("triphone (tri1, delta features)", lambda: (EXP_ROOT / "tri1" / "final.mdl").exists(), _train_tri1)


def _train_tri2():
    sh(f"steps/train_lda_mllt.sh --cmd run.pl "
       f"--beam {ALIGN_BEAM} --retry-beam {ALIGN_RETRY_BEAM} 2500 15000 "
       f"{TRAIN_DATA_DIR} {LANG_DIR} {EXP_ROOT}/tri1_ali {EXP_ROOT}/tri2", cwd=WORK_DIR)
    sh(f"steps/align_si.sh --cmd run.pl --nj {N_JOBS} "
       f"--beam {ALIGN_BEAM} --retry-beam {ALIGN_RETRY_BEAM} "
       f"{TRAIN_DATA_DIR} {LANG_DIR} {EXP_ROOT}/tri2 {EXP_ROOT}/tri2_ali", cwd=WORK_DIR)

stage("LDA+MLLT (tri2)", lambda: (EXP_ROOT / "tri2" / "final.mdl").exists(), _train_tri2)


def _train_tri3():
    sh(f"steps/train_sat.sh --cmd run.pl "
       f"--beam {ALIGN_BEAM} --retry-beam {ALIGN_RETRY_BEAM} 2500 15000 "
       f"{TRAIN_DATA_DIR} {LANG_DIR} {EXP_ROOT}/tri2_ali {EXP_ROOT}/tri3", cwd=WORK_DIR)

stage("SAT (tri3) -- final model", lambda: (EXP_ROOT / "tri3" / "final.mdl").exists(), _train_tri3)
```

## 14. Decode + Evaluate the Final Model (SAT, 2-gram, PS27)

Builds the decode graph from `lang_2g` and the `tri3` SAT model, decodes the
test set, and sweeps the LM weight over 1-25 before printing the best WER --
matches `train.ipynb` Section 12. Compare the printed number against
`conclusion.md`'s reported 5.41% WER for this exact configuration (expect a
gap -- this is a different, smaller corpus).

Graph building and decoding are each skipped if already done; the final
WER printout always re-runs (cheap) even when decoding itself was skipped,
so you always see the result.

```python
GRAPH_DIR = EXP_ROOT / "tri3" / "graph"
DECODE_DIR = EXP_ROOT / "tri3" / "decode_test"

def _mkgraph():
    sh(f"utils/mkgraph.sh {LANG_2G_DIR} {EXP_ROOT}/tri3 {GRAPH_DIR}", cwd=WORK_DIR)

stage("build decode graph (mkgraph)", lambda: (GRAPH_DIR / "HCLG.fst").exists(), _mkgraph)


def _decode():
    # Same long-utterance rationale as ALIGN_BEAM/ALIGN_RETRY_BEAM above --
    # decode_fmllr.sh's own defaults (beam=13.0, lattice-beam=6.0) are also
    # tuned for short utterances.
    sh(f"steps/decode_fmllr.sh --cmd run.pl --nj {N_JOBS} "
       f"--beam 30 --lattice-beam 15 "
       f"--min-lmwt 1 --max-lmwt 25 "
       f"{GRAPH_DIR} {DATA_ROOT}/test {DECODE_DIR}", cwd=WORK_DIR)

stage(
    "decode test set",
    lambda: DECODE_DIR.exists() and any(DECODE_DIR.glob("wer_*")),
    _decode,
)

# Cheap to recompute/print every time, even when decoding itself was skipped.
sh(f"grep WER {DECODE_DIR}/wer_* | utils/best_wer.sh", cwd=WORK_DIR)
```

## 15. Copy Final Results to Windows

Everything under `WORK_DIR` (`exp/`, `data/`, the Kaldi checkout) lives on
the WSL-native filesystem for the performance reasons described in
Section 2 -- it isn't automatically visible as a normal folder from
Windows Explorer the way a `/mnt/...` path would be. This step copies just
the final, useful artifacts -- the trained SAT model, its decode graph,
and the decode/WER output (`exp/tri3`) -- to `RESULTS_DIR` on the Windows
side, so they survive independent of this WSL distro and are easy to
browse from Windows. Nothing else (features, alignments, intermediate GMM
stages, per-utterance logs) is copied -- see `LOCAL_WSL_SETUP.md`'s
"Storage considerations" section.

```python
import shutil

FINAL_MODEL_RESULTS_DIR = RESULTS_DIR / "tri3"

def _copy_results():
    RESULTS_DIR.mkdir(parents=True, exist_ok=True)
    if shutil.which("rsync"):
        sh(f"rsync -a {EXP_ROOT}/tri3/ {FINAL_MODEL_RESULTS_DIR}/")
    else:
        sh(f"rm -rf {FINAL_MODEL_RESULTS_DIR} && cp -r {EXP_ROOT}/tri3 {FINAL_MODEL_RESULTS_DIR}")

stage(
    "copy final model + decode results to Windows",
    lambda: (FINAL_MODEL_RESULTS_DIR / "final.mdl").exists(),
    _copy_results,
)

print(f"Final model, graph, and decode/WER output available at: {FINAL_MODEL_RESULTS_DIR}")
```
